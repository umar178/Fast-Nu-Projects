### Task 1

```python
courses = ["C1", "C2", "C3", "C4"]
slots = [1, 2, 3, 4]

solution = {}

def is_valid(course, slot):
    if slot in solution.values():
        return False

    if course == "C1" and slot == 2:
        return False

    if course == "C3" and slot == 4:
        return False

    return True


def backtrack(index):
    if index == len(courses):
        return True

    course = courses[index]

    for slot in slots:
        if is_valid(course, slot):
            solution[course] = slot

            if backtrack(index + 1):
                return True

            del solution[course]

    return False


backtrack(0)

print("Class Timetable:")
for course in courses:
    print(course, "-> Slot", solution[course])
```

### Task 2

```python
grid = [
    [0, 0, 2, 0],
    [0, 0, 0, 3],
    [0, 3, 0, 0],
    [1, 0, 0, 0]
]

def is_valid(row, col, num):
    for i in range(4):
        if grid[row][i] == num:
            return False
    for i in range(4):
        if grid[i][col] == num:
            return False
    sr = (row // 2) * 2
    sc = (col // 2) * 2
    for i in range(sr, sr + 2):
        for j in range(sc, sc + 2):
            if grid[i][j] == num:
                return False
    return True

def solve():
    for row in range(4):
        for col in range(4):
            if grid[row][col] == 0:
                for num in range(1, 5):
                    if is_valid(row, col, num):
                        grid[row][col] = num
                        if solve():
                            return True
                        grid[row][col] = 0
                return False
    return True

solve()

for row in grid:
    print(row)
```

### Task 3

```python
exams = ["E1", "E2", "E3", "E4", "E5"]
slots = [1, 2, 3]
conflicts = [("E1", "E2"), ("E2", "E3"), ("E4", "E5")]
solution = {}
best_solution = None
min_late = 999

def is_valid(exam, slot):
    for a, b in conflicts:
        if exam == a and b in solution and solution[b] == slot:
            return False
        if exam == b and a in solution and solution[a] == slot:
            return False
    return True

def count_late():
    c = 0
    for e in solution:
        if solution[e] == 3:
            c += 1
    return c

def backtrack(index):
    global best_solution, min_late
    if index == len(exams):
        late = count_late()
        if late < min_late:
            min_late = late
            best_solution = solution.copy()
        return
    exam = exams[index]
    for slot in slots:
        if is_valid(exam, slot):
            solution[exam] = slot
            backtrack(index + 1)
            del solution[exam]

backtrack(0)

for exam in exams:
    print(exam, best_solution[exam])

print(min_late)
```
